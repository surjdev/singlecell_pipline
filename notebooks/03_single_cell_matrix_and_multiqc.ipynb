# 📊 Smart-seq2 Notebook 3: Count Matrix QC & MultiQC

This notebook walks through:
1. Loading the assembled **AnnData** single-cell object (`.h5ad`)
2. Calculating single-cell QC metrics (Total Counts, Detected Genes, Mitochondrial Ratio)
3. Aggregating all logs into an interactive **MultiQC** dashboard

```
import os
import sys
from pathlib import Path
sys.path.insert(0, str(Path('..').resolve()))

import anndata as ad
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.utils import load_config
from src.sc_qc import SmartSeq2QC

config = load_config('../config/pipeline_config.yaml')
adata = ad.read_h5ad('../data/counts/gene_cell_count_matrix.h5ad')
adata
```

### 1. Single-Cell Per-Cell QC Summary

```
adata.obs
```

### 2. QC Distributions (Total Counts & Detected Genes)

```
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
sns.scatterplot(data=adata.obs, x='total_counts', y='n_genes_by_counts', ax=axes[0], color='#9b2226', s=70)
axes[0].set_title('Total Counts vs. Detected Genes')
axes[0].set_xlabel('Total Read Counts')
axes[0].set_ylabel('Number of Detected Genes')

sns.boxplot(data=adata.obs[['total_counts', 'n_genes_by_counts']], ax=axes[1], palette='Pastel1')
axes[1].set_title('Distribution of Cell Metrics')
plt.tight_layout()
plt.show()
```

### 3. Generate MultiQC Dashboard

```
qc_engine = SmartSeq2QC(config)
qc_engine.run_multiqc()
print("MultiQC report compiled in reports/multiqc/multiqc_report.html")
```
